```python
import numpy as np
```

```python
x=np.arange(1,5)
print(x)

fig,axis=plt.subplots(1,2)


axis[0].plot(x,x**2,marker="o",markerfacecolor="red",color="blue")
axis[0].set_title("Square")
axis[0].set_xlabel("Numbers")
axis[0].set_ylabel("Square of numbers")

axis[1].plot(x,x**3,marker="o",markerfacecolor="blue",color="red")
axis[1].set_title("Cube")
axis[1].set_xlabel("Numbers")
axis[1].set_ylabel("Cube of numbers")

plt.subplots_adjust(wspace=0.4,hspace=0.6)

plt.show()
```

```python
days=np.arange(7)

apple = [150, 152, 153, 151, 155, 157, 156]
google = [2800, 2820, 2810, 2830, 2840, 2850, 2860]
amazon = [3400, 3420, 3410, 3430, 3440, 3450, 3460]
microsoft = [299, 301, 300, 302, 303, 305, 304]

fig,axis=plt.subplots(2,2,figsize=(10,8))

axis[0,0].plot(days,apple,marker='o',color="red")
axis[0,0].set_title("Apple Stock Price")
axis[0,0].grid(True)


axis[0,1].plot(days,google,marker='o',color="green")
axis[0,1].set_title("Google Stock Price")
axis[0,1].grid(True)

axis[1,0].plot(days,amazon,marker='o',color="blue")
axis[1,0].set_title("Amazon Stock Price")
axis[1,0].grid(True)

axis[1,1].plot(days,microsoft,marker='o',color="purple")
axis[1,1].set_title("microsoft Stock Price")
axis[1,1].grid(True)

for ax in axis[1, :]:
    ax.set_xlabel("Day")
for ax in axis[:, 0]:
    ax.set_ylabel("Price ($)")

plt.tight_layout()
plt.show()
```

```python
fig,axs=plt.subplots(2,1,sharex=True)

axs[0].plot([1,2,3],[3,2,1])
axs[1].plot([1,2,3],[1,2,3])
plt.show()
```

```python
fig, axs = plt.subplots(1, 2,sharey=True)

axs[0].plot([1,2,3], [3,2,1])
axs[1].plot([1,2,3], [1,2,3])

plt.show()
```

```python
x=np.arange(10)
y=x**2
print(y)

plt.plot(x,y,linestyle=":",marker="o",color="green",markerfacecolor="red",linewidth=2,label='Square of Numbers')

plt.xlabel("X-axis",rotation=90)
plt.ylabel("Y-axis",rotation=45)
plt.grid()

plt.xticks(x)
plt.legend()
plt.show()
```

```python
x=np.arange(10)
y=x**2
print(y)

plt.plot(x,y,linestyle=":",marker="o",color="green",markerfacecolor="red",linewidth=2,label='Square of Numbers')

plt.xlabel("X-axis",rotation=90)
plt.ylabel("Y-axis",rotation=45)
plt.grid()

plt.xticks(x)
plt.yticks(y)
plt.legend()
plt.show()
```
